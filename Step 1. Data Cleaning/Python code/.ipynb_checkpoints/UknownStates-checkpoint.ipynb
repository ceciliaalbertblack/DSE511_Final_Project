```python
import pandas as pd

# 1. Load the CSV file
df = pd.read_csv('NORS_20251104.csv')

# 2. Inspect the data to confirm column name and data types
print("Data head and info:")
print(df.head())
print(df.info())

# 3. Count the number of unique states in the 'State' column
# This assumes your column is exactly named 'State'
# The result will be saved in the 'unique_states_count' variable
unique_states_count = df['State'].nunique()
print(f"\nThe number of different states is: {unique_states_count}")

# Get the actual list of unique states
unique_states_list = df['State'].unique()

sorted_states = sorted(unique_states_list)

print("List of Unique States (Alphabetical):")
print(sorted_states)
```

```python
import pandas as pd

# Load the CSV file
df = pd.read_csv('NORS_20251104.csv', low_memory=False)

# 1. Total records
total_records = len(df)

# 2. List of non-state jurisdictions
non_state_jurisdictions = [
    'District of Columbia',
    'Guam',
    'Multistate',
    'Northern Mariana Islands',
    'Puerto Rico',
    'Republic of Palau',
    'Republic of the Marshall Islands',
    'Virgin Islands'
]

# 3. Get the individual counts
non_state_counts_series = df[df['State'].isin(non_state_jurisdictions)]['State'].value_counts()

# 4. Convert the Series to a DataFrame for calculation
result_df = non_state_counts_series.to_frame(name='Count')

# 5. Calculate the percentage
result_df['Percentage'] = (result_df['Count'] / total_records) * 100

# 6. Format the percentage column
result_df['Percentage'] = result_df['Percentage'].map('{:.4f}%'.format)

print("\n--- Non-State Jurisdiction Counts and Percentages ---")
print(result_df)
print(f"\nTotal Records in Dataset: {total_records}")
```
